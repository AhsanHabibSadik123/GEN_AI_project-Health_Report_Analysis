```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)
```

```python
with open("blood_work.txt", "r") as f:
    blood_work_data = f.read()
```

```python
blood_work_data
```

```python
extraction_prompt = f"""
You are a medical data extraction assistant.
from the medical data report bellow, EXTRACT all the test values and classify each one as HIGH, LOW, or NORMAL based on the reference ranges provided in the report.
Fromat your response as:
- Test Name: Value | Status (HIGH, LOW, NORMAL) | Reference Range

Medical Data Report:
{blood_work_data}
"""
```

```python
extraction_prompt
```

```python
extracted_data = llm.invoke(extraction_prompt)
print(extracted_data.content)
```

```python
diet_prompt = f"""
you are a nutritionist specialized in Bangadeshi dietary habits. 
Based on the blood work data provided, Write:
1. A short health summary in 3 lines explaining the current health condition in simple language. 
2. A short Bangladeshi diet plan having only two sections (1) Foods to eat more of, (2) Foods to avoid.
Do not include any other sections in diet plan.
blood work data:
{extracted_data.content}
"""
```

```python
diet_prompt
```

```python
diet_plan = llm.invoke(diet_prompt)
print(diet_plan.content)
```
